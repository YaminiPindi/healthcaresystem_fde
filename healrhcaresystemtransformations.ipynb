```python
from pyspark.sql.functions import col, explode, lit

# First explode the entry array to get individual observations
df_exploded = df.select(explode(col("entry")).alias("entry"))

# Now flatten the observation fields from entry.resource
df_flat = df_exploded.select(
    col("entry.resource.resourceType"),
    col("entry.resource.id").alias("observation_id"),
    col("entry.resource.meta.lastUpdated").alias("meta_last_updated"),
    col("entry.resource.meta.tag")[0]["system"].alias("meta_tag_system"),
    col("entry.resource.meta.tag")[0]["code"].alias("meta_tag_code"),
    col("entry.resource.status"),
    col("entry.resource.category")[0]["coding"][0]["system"].alias("category_system"),
    col("entry.resource.category")[0]["coding"][0]["code"].alias("category_code"),
    col("entry.resource.category")[0]["coding"][0]["display"].alias("category_display"),
    col("entry.resource.code.coding")[0]["system"].alias("code_system"),
    col("entry.resource.code.coding")[0]["code"].alias("code"),
    col("entry.resource.code.coding")[0]["display"].alias("code_display"),
    col("entry.resource.code.text").alias("code_text"),
    col("entry.resource.subject.reference").alias("subject_reference"),
    col("entry.resource.encounter.reference").alias("encounter_reference"),
    col("entry.resource.effectiveDateTime"),
    col("entry.resource.issued"),
    col("entry.resource.valueQuantity.value").alias("value"),
    col("entry.resource.valueQuantity.unit").alias("value_unit"),
    col("entry.resource.valueQuantity.system").alias("value_system"),
    col("entry.resource.valueQuantity.code").alias("value_code"),
    col("entry.search.mode").alias("search_mode"),
    col("entry.response.status").alias("response_status"),
    col("entry.response.etag").alias("response_etag")
)

display(df_flat)
```

```python
from pyspark.sql.functions import regexp_replace, regexp_extract, to_timestamp, col

# Step 1: Remove unnecessary columns
df_observations = df_flat.drop(
    "response_etag",
    "response_status",
    "search_mode",
    "value_code",
    "effectiveDateTime",
    "code_text",
    "code_display",
    "meta_tag_system",
    "value_system",
    "code_system",
    "category_code",
    "category_system"
)

# Step 2: Apply data type conversions
df_observations = df_observations \
    .withColumn(
        "subject_reference",
        regexp_replace(col("subject_reference"), "Patient/", "")
    ) \
    .withColumn(
        "encounter_reference",
        regexp_replace(col("encounter_reference"), "Encounter/", "")
    ) \
    .withColumn(
        "meta_last_updated",
        to_timestamp(col("meta_last_updated"))
    ) \
    .withColumn(
        "issued",
        to_timestamp(col("issued"))
    )

print("="*80)
print("DATA TRANSFORMATIONS APPLIED")
print("="*80)
print(f"\nTotal observations: {df_observations.count()}")
print(f"Total columns: {len(df_observations.columns)}")
print(f"\nColumns: {', '.join(df_observations.columns)}")
print("\nData Types:")
for col_name, col_type in df_observations.dtypes:
    print(f"  - {col_name}: {col_type}")
print("="*80)

display(df_observations)
```

```python
from pyspark.sql.functions import row_number, col
from pyspark.sql.window import Window

print("="*80)
print("CREATING STAR SCHEMA: DIMENSION AND FACT TABLES")
print("="*80)

# Dimension Table 1: Patient (PRIMARY KEY: patient_id)
print("\n1. Creating dim_patient...")
dim_patient = df_observations.select(
    col("subject_reference").alias("patient_id")
).distinct()

# Add surrogate key using row_number for sequential IDs
window_patient = Window.orderBy("patient_id")
dim_patient = dim_patient.withColumn("patient_key", row_number().over(window_patient))
dim_patient = dim_patient.select("patient_key", "patient_id")

print(f"   ✅ dim_patient created: {dim_patient.count()} patients")
print(f"   Primary Key: patient_id (string UUID)")
print(f"   Surrogate Key: patient_key (integer)")

# Dimension Table 2: Encounter (PRIMARY KEY: encounter_id)
print("\n2. Creating dim_encounter...")
dim_encounter = df_observations.select(
    col("encounter_reference").alias("encounter_id")
).distinct()

window_encounter = Window.orderBy("encounter_id")
dim_encounter = dim_encounter.withColumn("encounter_key", row_number().over(window_encounter))
dim_encounter = dim_encounter.select("encounter_key", "encounter_id")

print(f"   ✅ dim_encounter created: {dim_encounter.count()} encounters")
print(f"   Primary Key: encounter_id (string UUID)")
print(f"   Surrogate Key: encounter_key (integer)")

# Dimension Table 3: Observation Code (PRIMARY KEY: code)
print("\n3. Creating dim_observation_code...")
dim_observation_code = df_observations.select(
    col("code")
).distinct()

window_code = Window.orderBy("code")
dim_observation_code = dim_observation_code.withColumn("observation_code_key", row_number().over(window_code))
dim_observation_code = dim_observation_code.select("observation_code_key", "code")

print(f"   ✅ dim_observation_code created: {dim_observation_code.count()} unique LOINC codes")
print(f"   Primary Key: code (string LOINC code)")
print(f"   Surrogate Key: observation_code_key (integer)")

# Dimension Table 4: Category (PRIMARY KEY: category_display)
print("\n4. Creating dim_category...")
dim_category = df_observations.select(
    col("category_display")
).distinct()

window_category = Window.orderBy("category_display")
dim_category = dim_category.withColumn("category_key", row_number().over(window_category))
dim_category = dim_category.select("category_key", "category_display")

print(f"   ✅ dim_category created: {dim_category.count()} categories")
print(f"   Primary Key: category_display (string)")
print(f"   Surrogate Key: category_key (integer)")

# Fact Table: Observation (optimized with only measures + foreign keys)
print("\n5. Creating fact_observation...")
fact_observation = df_observations \
    .join(dim_patient, df_observations.subject_reference == dim_patient.patient_id, "left") \
    .join(dim_encounter, df_observations.encounter_reference == dim_encounter.encounter_id, "left") \
    .join(dim_observation_code, df_observations.code == dim_observation_code.code, "left") \
    .join(dim_category, df_observations.category_display == dim_category.category_display, "left") \
    .select(
        col("observation_id"),
        col("patient_key"),
        col("encounter_key"),
        col("observation_code_key"),
        col("category_key"),
        col("meta_last_updated"),
        col("issued"),
        col("value"),
        col("value_unit"),
        col("status")
    )

print(f"   ✅ fact_observation created: {fact_observation.count()} observations")
print(f"   Primary Key: observation_id")
print(f"   Foreign Keys: patient_key, encounter_key, observation_code_key, category_key")
print(f"   Measures: value, value_unit")
print(f"   Attributes: meta_last_updated, issued, status")

print("\n" + "="*80)
print("STAR SCHEMA CREATED SUCCESSFULLY")
print("="*80)
print(f"\nDimension Tables: 4")
print(f"  - dim_patient: {dim_patient.count()} rows")
print(f"  - dim_encounter: {dim_encounter.count()} rows")
print(f"  - dim_observation_code: {dim_observation_code.count()} rows")
print(f"  - dim_category: {dim_category.count()} rows")
print(f"\nFact Table: 1")
print(f"  - fact_observation: {fact_observation.count()} rows")
print("="*80)
```

```python
print("="*80)
print("PRIMARY KEY VALIDATION")
print("="*80)

def validate_primary_key(df, pk_column, table_name):
    """Validate that a column meets primary key constraints (non-null, unique)"""
    total_rows = df.count()
    null_count = df.filter(col(pk_column).isNull()).count()
    distinct_count = df.select(pk_column).distinct().count()
    
    is_valid = (null_count == 0) and (total_rows == distinct_count)
    
    print(f"\n{table_name}:")
    print(f"  Primary Key: {pk_column}")
    print(f"  Total rows: {total_rows}")
    print(f"  Null values: {null_count} {'✅' if null_count == 0 else '❌'}")
    print(f"  Distinct values: {distinct_count}")
    print(f"  Duplicates: {total_rows - distinct_count} {'✅' if total_rows == distinct_count else '❌'}")
    print(f"  Status: {'✅ VALID PRIMARY KEY' if is_valid else '❌ INVALID PRIMARY KEY'}")
    
    return is_valid

# Validate all primary keys
all_valid = True
all_valid &= validate_primary_key(dim_patient, "patient_id", "dim_patient")
all_valid &= validate_primary_key(dim_encounter, "encounter_id", "dim_encounter")
all_valid &= validate_primary_key(dim_observation_code, "code", "dim_observation_code")
all_valid &= validate_primary_key(dim_category, "category_display", "dim_category")
all_valid &= validate_primary_key(fact_observation, "observation_id", "fact_observation")

print("\n" + "="*80)
if all_valid:
    print("✅ ALL PRIMARY KEYS ARE VALID")
else:
    print("❌ SOME PRIMARY KEYS ARE INVALID")
print("="*80)
```

```python
print("="*80)
print("SAVING TABLES TO SILVER LAYER")
print("="*80)

# Ensure storage account key is configured
spark.conf.set(
    "fs.azure.account.key.healthcaresystemstoreage.dfs.core.windows.net",
    "<YOUR_AZURE_STORAGE_ACCOUNT_KEY>"
)

# Define silver layer base path
silver_base_path = "abfss://silver@healthcaresystemstoreage.dfs.core.windows.net"

# Save dimension and fact tables
print("\nSaving Tables...")

tables_to_save = [
    (dim_patient, "dim_patient"),
    (dim_encounter, "dim_encounter"),
    (dim_observation_code, "dim_observation_code"),
    (dim_category, "dim_category"),
    (fact_observation, "fact_observation")
]

for df_table, table_name in tables_to_save:
    table_path = f"{silver_base_path}/{table_name}"
    print(f"   Saving {table_name}...")
    
    df_table.write.format("delta") \
        .mode("overwrite") \
        .option("overwriteSchema", "true") \
        .save(table_path)
    
    row_count = df_table.count()
    print(f"   ✅ {table_name} saved: {row_count} rows")

print("\n" + "="*80)
print("✅ ALL TABLES SAVED TO SILVER LAYER SUCCESSFULLY")
print("="*80)
print(f"\nSilver Layer Location: {silver_base_path}")
print(f"\nTables Saved:")
print(f"  1. dim_patient ({dim_patient.count()} rows)")
print(f"  2. dim_encounter ({dim_encounter.count()} rows)")
print(f"  3. dim_observation_code ({dim_observation_code.count()} rows)")
print(f"  4. dim_category ({dim_category.count()} rows)")
print(f"  5. fact_observation ({fact_observation.count()} rows)")
print("\nFormat: Delta")
print("Mode: Overwrite")
print("Schema: Overwrite enabled")
print("="*80)
```

```python
from pyspark.sql.functions import (
    col, count, countDistinct, min, max, avg, round as spark_round,
    first, collect_set, date_format, year, month, dayofweek,
    sum as spark_sum, when, concat_ws
)

print("="*80)
print("GOLD LAYER: CREATING BUSINESS-READY ANALYTICS TABLES")
print("="*80)

# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# GOLD TABLE 1: Fully Denormalized Observations
# Purpose: Single wide table for ad-hoc analytics — no joins needed
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
print("\n1. Creating gold_observations_enriched...")

gold_observations_enriched = fact_observation \
    .join(dim_patient, "patient_key", "left") \
    .join(dim_encounter, "encounter_key", "left") \
    .join(dim_observation_code, "observation_code_key", "left") \
    .join(dim_category, "category_key", "left") \
    .select(
        col("observation_id"),
        col("patient_id"),
        col("encounter_id"),
        col("code").alias("loinc_code"),
        col("category_display").alias("category"),
        col("status"),
        col("value").alias("observation_value"),
        col("value_unit"),
        col("issued").alias("observation_datetime"),
        col("meta_last_updated"),
        year(col("issued")).alias("observation_year"),
        month(col("issued")).alias("observation_month"),
        dayofweek(col("issued")).alias("observation_dayofweek"),
        date_format(col("issued"), "yyyy-MM").alias("observation_year_month")
    )

obs_count = gold_observations_enriched.count()
print(f"   ✅ gold_observations_enriched: {obs_count} rows, {len(gold_observations_enriched.columns)} columns")

# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# GOLD TABLE 2: Patient Summary
# Purpose: One row per patient with aggregated metrics
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
print("\n2. Creating gold_patient_summary...")

gold_patient_summary = gold_observations_enriched.groupBy("patient_id").agg(
    count("observation_id").alias("total_observations"),
    countDistinct("encounter_id").alias("total_encounters"),
    countDistinct("loinc_code").alias("distinct_test_types"),
    min("observation_datetime").alias("first_observation_date"),
    max("observation_datetime").alias("last_observation_date"),
    spark_round(avg("observation_value"), 2).alias("avg_observation_value"),
    concat_ws(", ", collect_set("category")).alias("categories_observed"),
    spark_sum(when(col("category") == "vital-signs", 1).otherwise(0)).alias("vital_signs_count"),
    spark_sum(when(col("category") == "laboratory", 1).otherwise(0)).alias("lab_count"),
    spark_sum(when(col("category") == "survey", 1).otherwise(0)).alias("survey_count"),
    spark_sum(when(col("category") == "procedure", 1).otherwise(0)).alias("procedure_count")
)

patient_count = gold_patient_summary.count()
print(f"   ✅ gold_patient_summary: {patient_count} rows, {len(gold_patient_summary.columns)} columns")

# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# GOLD TABLE 3: Encounter Summary
# Purpose: One row per encounter with aggregated observation details
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
print("\n3. Creating gold_encounter_summary...")

gold_encounter_summary = gold_observations_enriched.groupBy("encounter_id", "patient_id").agg(
    count("observation_id").alias("total_observations"),
    countDistinct("loinc_code").alias("distinct_test_types"),
    min("observation_datetime").alias("encounter_start"),
    max("observation_datetime").alias("encounter_end"),
    concat_ws(", ", collect_set("category")).alias("categories"),
    concat_ws(", ", collect_set("loinc_code")).alias("loinc_codes_tested"),
    spark_sum(when(col("status") == "final", 1).otherwise(0)).alias("final_count"),
    spark_sum(when(col("status") != "final", 1).otherwise(0)).alias("non_final_count")
)

enc_count = gold_encounter_summary.count()
print(f"   ✅ gold_encounter_summary: {enc_count} rows, {len(gold_encounter_summary.columns)} columns")

# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# WRITE ALL GOLD TABLES TO ADLS GOLD CONTAINER
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
print("\n" + "="*80)
print("WRITING GOLD TABLES TO ADLS")
print("="*80)

gold_base_path = "abfss://gold@healthcaresystemstoreage.dfs.core.windows.net"

gold_tables = [
    (gold_observations_enriched, "gold_observations_enriched"),
    (gold_patient_summary, "gold_patient_summary"),
    (gold_encounter_summary, "gold_encounter_summary")
]

for df_gold, table_name in gold_tables:
    table_path = f"{gold_base_path}/{table_name}"
    print(f"\n   Writing {table_name} → {table_path}")
    df_gold.write.format("delta") \
        .mode("overwrite") \
        .option("overwriteSchema", "true") \
        .save(table_path)
    print(f"   ✅ {table_name} saved successfully")

print("\n" + "="*80)
print("✅ GOLD LAYER COMPLETE")
print("="*80)
print(f"\nGold Layer Location: {gold_base_path}")
print(f"\nTables Written:")
print(f"  1. gold_observations_enriched  ({obs_count} rows) — Denormalized observations")
print(f"  2. gold_patient_summary         ({patient_count} rows) — Patient-level aggregations")
print(f"  3. gold_encounter_summary       ({enc_count} rows) — Encounter-level aggregations")
print(f"\nFormat: Delta | Mode: Overwrite")
print("="*80)

# Display sample data
print("\n📊 SAMPLE: gold_observations_enriched")
display(gold_observations_enriched.limit(10))

print("\n📊 SAMPLE: gold_patient_summary")
display(gold_patient_summary.limit(10))

print("\n📊 SAMPLE: gold_encounter_summary")
display(gold_encounter_summary.limit(10))
```
